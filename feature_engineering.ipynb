```python
import pandas as pd
import numpy as np
```

```python
# Load cleaned data
file_path = r"C:\Users\Vanitha\OneDrive\Documents\GUVI\capstone_project\job_acceptance_prediction_system\data\cleaned\job_acceptance_cleaned.csv"
df = pd.read_csv(file_path)

df.shape
```

```python
# 1️. Academic Band
# Based on degree percentage
def academic_band(deg):
    if deg >= 75:
        return "High"
    elif deg >= 60:
        return "Medium"
    else:
        return "Low"

df["academic_band"] = df["degree_percentage"].apply(academic_band) 
```

```python
# 2️. Experience Category
def experience_category(exp):
    if exp < 2:
        return "Fresher"
    elif exp < 5:
        return "Junior"
    else:
        return "Senior"

df["experience_category"] = df["years_of_experience"].apply(experience_category) 
```

```python
# 3️. Skills Match Level
df["skills_match_level"] = pd.cut(
    df["skills_match_percentage"],
    bins=[0, 40, 70, 100],
    labels=["Low", "Medium", "High"]
) 
```

```python
# 4️. Interview Score Average
df["interview_score_avg"] = (
    df["technical_score"]
    + df["aptitude_score"]
    + df["communication_score"]
) / 3 
```

```python
# 5️. Interview Performance Category
def interview_performance(score):
    if score >= 75:
        return "Excellent"
    elif score >= 50:
        return "Average"
    else:
        return "Poor"

df["interview_performance"] = df["interview_score_avg"].apply(interview_performance) 
```

```python
# 6️. Placement Probability Score (analytical feature)
df["placement_probability_score"] = df["skills_match_percentage"] / 100 
```

```python
# 7️. CTC Gap
df["ctc_gap"] = df["expected_ctc_lpa"] - df["previous_ctc_lpa"] 
```

```python
# 8️. CTC Gap Flag
df["ctc_gap_flag"] = df["ctc_gap"].apply(lambda x: 1 if x > 5 else 0) 
```

```python
# Select FINAL features
final_features = [
    # Raw selected features
    "ssc_percentage",
    "hsc_percentage",
    "degree_percentage",
    "technical_score",
    "aptitude_score",
    "communication_score",
    "skills_match_percentage",
    "certifications_count",
    "years_of_experience",
    "previous_ctc_lpa",
    "expected_ctc_lpa",
    "notice_period_days",
    "employment_gap_months",
    "company_tier",
    "relevant_experience",
    "internship_experience",

    # Engineered features
    "academic_band",
    "experience_category",
    "skills_match_level",
    "interview_score_avg",
    "interview_performance",
    "placement_probability_score",
    "ctc_gap",
    "ctc_gap_flag",

    # Target
    "status"
]

model_df = df[final_features]
```
